#For - Descoberta guiada

Crie uma lista com os valores 11,22,51,86,12,71 e 63. Calcule o valor da soma total da lista e imprima na tela.

```python
lista_numeros = [11, 22, 51, 86, 12, 71, 63]

soma_numeros = 0
menor_numero = lista_numeros[0]

for numero in lista_numeros:
  soma_numeros = soma_numeros + numero
  if numero < menor_numero:
    menor_numero = numero

print("A soma dos numeros é: ",soma_numeros)
print("A média é: ", soma_numeros / len(lista_numeros))
print("O menor numero é: ", menor_numero)
```

#While - Descoberta guiada

Crie uma lista que some os valores de input do usuário até que seja digitado 0. Quando o 0 for digitado, o algoritmo deve finalizar e apresentar o total.

```python
lista = []
entrada = ""
while entrada != "0":
  entrada = input("Informe um elemnto novo: ")
  if entrada != "0":
    lista.append(int(entrada))
    print(lista)

soma = 0
posicao = 0
while posicao < len(lista):
  print("Soma inicio ", soma)
  print("Posicao inicio", posicao)
  soma = soma + lista[posicao]
  posicao = posicao + 1
  print("Soma ", soma)
  print("Posicao ", posicao)

print("Soma :", soma)
```

```python
def jogo_do_pim(multiplo, limite):
  contador = 1
  resultado_do_pim = ""

  while contador <= limite:
    resto_da_divisao = contador % multiplo
    if resto_da_divisao == 0:
      resultado_do_pim = resultado_do_pim + "Pim" + "\n"
    else:
      resultado_do_pim = resultado_do_pim + str(contador) + "\n"
    contador = contador + 1

  return resultado_do_pim

print(jogo_do_pim(2,10))
print(jogo_do_pim(3,9))
```
